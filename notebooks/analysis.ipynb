```python
import matplotlib.pyplot as plt
from qem_bench import results

df = results.load("ghz_baseline")
df = df.sort_values("timestamp").drop_duplicates(["circuit", "n", "technique"], keep="last")
df
```

```python
plt.errorbar(df["n"], df["estimate"], yerr=df["std"], marker="o", capsize=3, label="No mitigation")
plt.axhline(1, linestyle="--", color="grey", label="Ideal")
plt.xlabel("Number of qubits")
plt.ylabel("⟨X⊗X⊗…⊗X⟩")
plt.title("GHZ on simulated ibm_fez")
plt.ylim(0, 1.1)
plt.legend()
plt.savefig(results.RESULTS_DIR / "ghz_baseline.png", dpi=150)
```
